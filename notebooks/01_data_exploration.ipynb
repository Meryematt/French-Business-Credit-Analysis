```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    "../data/raw/credits_entreprises.csv",
    sep=";",
    decimal=",",
    encoding="utf-8-sig",
    low_memory=False
)

# df.head()
# df.shape
# df.columns
# df.info()
# df.describe()

print(df["series_key"].nunique())
print(df["series_key"].value_counts())
print(df.groupby("series_key")["obs_value"].mean())
print(df.groupby("series_key")["obs_value"].count())

# stats = df.groupby("series_key")["obs_value"].agg(
#     ["mean", "std"]
# )

# stats["cv"] = (stats["std"] / stats["mean"]) * 100

# print(stats)

series_analysis = df.groupby("series_key").agg(
    total_rows=("obs_value", "size"),
    available_values=("obs_value", "count"),
    missing_values=("obs_value", lambda x: x.isna().sum())
)

series_analysis["missing_percentage"] = (
    series_analysis["missing_values"] /
    series_analysis["total_rows"] * 100
)

print(series_analysis)

series_analysis.sort_values(
    by="missing_percentage",
    ascending=False
).head(10)

print(df.duplicated().sum())
print(df[["series_key", "time_period"]].duplicated().sum())
```
